```python
import json
```

```python
d = {'id':12324,'status':'€ : connecting 👋'}
print(d)
print(type(d))
print(len(d))
```

# Dict -> String = Serialization

```python
string = json.dumps(d)
print(string)
print(type(string))
print(len(string))
```

# String -> Bytes = Encoding

- UTF-8 = Unicode Transformation Format - 8 bit

```python
encoded_data = string.encode('utf-8')
print(encoded_data)
print(type(encoded_data))
print(len(encoded_data))
```

### Encoding without ascii escape sequence

```python
string = json.dumps(d,ensure_ascii=False)
print(string)
print(type(string))
print(len(string))
```

```python
encoded_data = string.encode('utf-8')
print(encoded_data)
print(type(encoded_data))
print(len(encoded_data))
```
