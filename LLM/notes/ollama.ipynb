## Ollama Library

```python
import ollama
```

```python
response = ollama.generate(model='gemma3:4b',prompt='why does moon glow')
response
```

```python
print(response.model)
print(response.created_at)
print(response.done_reason)
print(response.response)
```

```python
response = ollama.generate(model='gemma3:4b',prompt='why does moon glow',stream=True)
for i in   response:
    print(i.response , end ='')
```

# Rest APIs

- Ollama internally work using REST API hosted locally at port 11434
- CLI Commands and Python library are just wrappers around the API endpoints.

```python
import requests
```

```python
ollama_url = "http://localhost:11434/api/generate"

payload = {
    'model' : 'gemma3:4b',
    'prompt' : 'hi'
}

resp = requests.post(ollama_url,json=payload)
print(resp)
```

```python
for i in resp.iter_lines():
    print(i)
```

```python
import json
output = ''

for i in resp.iter_lines():
    if i:
        data = json.loads(i.decode())
        if 'response' in data:
            output += data['response']
        if data.get('done'):
            break
        
print(output)        
        
```

```python
for i in resp.iter_content():
    print(i, end='')
```
