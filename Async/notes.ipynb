# Concurrency Programming

- Concurrency is essentially applicable when at least two tasks can start run and complete in overlapping time.
- Programming of such tasks are called concurrency programming.
- It doesn’t necessarily mean, that they’ll ever be running at the same instant (eg. multiple threads on a single-core machine)

# Event Loops

- Think of the Event Loop as the "manager" or the "beating heart" of an asynchronous Python program. It is the invisible engine that schedules tasks, pauses them when they hit an await, and switches to other work.
- In a normal Python script, code runs top-to-bottom. If a line takes 10 seconds, the whole script sits there for 10 seconds.
- The Event Loop changes this. It maintains a Task Queue. It picks a task, runs it until that task says "I'm waiting for something" (via await), then puts that task aside and grabs the next one from the queue.
- How it operates:
    - Monitor: It constantly checks if any "waiting" tasks (like a network request or a timer) are finished.
    - Execute: If a task is ready, the loop resumes it.
    - Idle: If all tasks are waiting, the loop itself pauses to save CPU until something happens.


- Before Python 3.7, setting up the event loop was manual and clunky. asyncio.run() was introduced as the "one-stop-shop" to start your async program.

    - When you call asyncio.run(main()), three major things happen:

    - Creates the Loop: It spins up a brand-new Event Loop.

    - Runs the Coroutine: It takes your main() coroutine and runs it until it's finished.

    - Clean up: Once main() is done, it shuts down the loop, closes any remaining "generators," and cleans up the thread so other code can run normally.

# Asynchronous Programming

- You can write you asynchronous code in three ways:
    - Using Futures, not used generally in python, used for level coding
    - Using Coroutines
    - Wrapping Coroutines as Tasks

- If you just await your functions without wrapping it as task you won't get proper concurrency.
- tasks will be scheduled in event loop so that if a task is waiting, another task can be started or resumed.
- in other cases coroutines can't be executed concurrently since waiting of one coroutine will make the entire event loop waiting.

# Coroutine

- if await is the action of waiting, then Coroutines and Tasks are the "things" you are actually waiting for. They are the building blocks of asynchronous Python, but they serve different purposes.
- A Coroutine is a function that can be paused and resumed. When you define a function with async def, calling it doesn't actually run the code inside; it just creates a coroutine object (essentially a "plan" for what to do)
- An async def function object.
- Starts only when awaited.
- Sequential (one by one).
- You wait for it to finish.
- When you call a async funtion it won't run the code inside, it returns a coroutine object which can awaited.

# Tasks

- A Task is a wrapper for a coroutine that schedules it to run on the Event Loop as soon as possible. Once a coroutine is wrapped in a Task, it starts running in the background independently of the current function's flow.
- object.	A scheduled "background job" wrapping a coroutine.
- Starts immediately upon creation.
- Concurrent (multiple at once).
- You can cancel it or check its status.
